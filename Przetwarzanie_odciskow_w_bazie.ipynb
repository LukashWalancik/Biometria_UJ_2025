# Przetwarzanie odcisków w bazie danych dla późniejszej identyfikacji nowych odcisków

(Do zrobienia na początku i po dodaniu nowych odcisków)

```python
import json
import os
from pathlib import Path
from utils import proces_image_and_get_points

fingerprints = {}

dir_path_str = 'Database/all'
dir_path = Path(dir_path_str)
files = os.listdir(dir_path)
for file in files:
    points = proces_image_and_get_points(Path(dir_path_str + '/' + file))
    raw_filename = file.split('.')[0]
    fingerprints[raw_filename] = points

with open(Path('Database/processed/processed_data.json'), 'w') as f:
    f.write(json.dumps(fingerprints))

print("Przetworzono wszytskie odciski i zapisano ich dane.")
print(fingerprints)
```
